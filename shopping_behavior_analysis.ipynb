```python
df["Review Rating"].fillna(df.groupby("Category")["Review Rating"].transform("median"), inplace = True)
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
```

```python
df = df.rename(columns = {"purchase_amount_(usd)":"purchase_amount"})
```

```python
#create a new column age group

labels = ["young_adult","adult","middle_aged","senior"]
df["age_group"] = pd.qcut(df["age"],4,labels=labels)
```

```python
df.head()
```

```python
#purchase frequency days columns

frequency = {
    "Fortnightly" : 14,
    "Weekly" : 7,
    "Monthly" : 30,
    "Quarterly" : 90,
    "Bi-Weekly" : 14,
    "Annually" : 365,
    "Every 3 Months" : 90
}

df["purchase_frequency_days"] = df["frequency_of_purchases"].map(frequency)


```
